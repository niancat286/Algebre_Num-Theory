### 1. Абелеві групи




1. Напишіть функцію, яка для заданого $n$ повертає кількість попарно неізоморфних абелевих груп порядку $n$.
2. Знайдіть перші сто значень $n$, для яких існує рівно одна абелева група порядку $n$, і порівняйте з розкладом $n$ у добуток простих.
3. Напишіть гіпотезу про те, для яких $n$ існує рівно одна абелева група порядку $n$.
4. Спробуйте знайти всі $n$, для яких існує рівно $n$ абелевих груп порядку $n$.



```python
############################################################
# Пункт 1.
# Функція, яка для заданого n повертає кількість
# попарно неізоморфних абелевих груп порядку n.
############################################################

# функція числа розбиттів p(k)
def partition_number(k):
    dp = [0] * (k + 1)
    dp[0] = 1
    for i in range(1, k + 1):
        for j in range(i, k + 1):
            dp[j] += dp[j - i]
    return dp[k]

#  факторизація
def factorize(n):
    p = 2
    factors = []
    while p * p <= n:
        e = 0
        while n % p == 0:
            e += 1
            n //= p
        if e > 0:
            factors.append((p, e))
        p += 1 if p == 2 else 2
    if n > 1:
        factors.append((n, 1))
    return factors

# Основна функція з пункту 1
def num_abelian_groups(n):
    facts = factorize(n)
    result = 1
    for (p, e) in facts:
        result *= partition_number(e)
    return result


############################################################
# Пункт 2.
# Перші 100 n, для яких існує рівно 1 абелева група порядку n.
# (тобто num_abelian_groups(n) == 1)
# + вивести розклад n на прості.
############################################################

first_100 = []
x = 1
while len(first_100) < 100:
    if num_abelian_groups(x) == 1:
        first_100.append(x)
    x += 1

print("ПЕРШІ 100 n, для яких a(n)=1:")
for n in first_100:
    print(n, " -> ", factorize(n))


############################################################
# Пункт 3.
# Гіпотеза.
############################################################

print("\nГІПОТЕЗА:")
print("a(n)=1 тоді й тільки тоді, коли n — squarefree (без квадратів простих).")


############################################################
# Пункт 4.
# Знайти всі n (у межі 1..20000), для яких a(n)=n.
############################################################

solutions = []
for n in range(1, 20001):
    if num_abelian_groups(n) == n:
        solutions.append(n)

print("\nВСІ n до 20000, для яких a(n)=n:")
print(solutions)
```

### 2. Дільники нуля в кільцях

### 3. Корінь з одиниці за модулем $p$

1. Знайдіть перші сто простих чисел $p$, для яких факторкільце $\mathbb{Z}_p[x]/(x^2+1)$ є полем.  $\qquad$ % PolynomialRing(GF(p))
2. Знайдіть остачі при ділення цих простих чисел на $m$ для деяких значень $m=2,3,\ldots,10$.        
3. Спробуйте знайти закономірність і сформулювати гіпотезу про такі прості числа.


```python
# 1. Знаходження перших 100 простих чисел p конгруентних 3 (mod 4)
primes_list = []
count = 0
p = 3  # Починаємо з першого простого, що задовольняє умові

print("--- 1. Знаходження простих чисел p конгруентних 3 (mod 4) ---")

while count < 100:
    # is_prime(p) - вбудована функція SageMath
    if is_prime(p) and (p % 4 == 3):
        primes_list.append(p)
        count += 1
    
    # Оскільки p має бути непарним, можна збільшувати крок для прискорення
    if p == 3:
        p = 5 # Наступне непарне
    else:
        p += 2

last_p = primes_list[-1]
print(f"Знайдено {len(primes_list)} простих чисел.")
print(f"Перші 5: {primes_list[:5]}")
print(f"100-е просте число, що задовольняє умові: {last_p}")
print("-" * 60)

# 2. Обчислення остач для m = 2, 3, ..., 10

mods = range(2, 11)
results = {}

print("--- 2. Обчислення остач p mod m ---")

for m in mods:
    remainders = set()
    for p in primes_list:
        remainders.add(p % m)
    results[m] = sorted(list(remainders))
    
    # Виведення результатів для кожного m
    print(f"Остачі p mod {m}: {results[m]}")

print("-" * 60)

# 3. Формулювання гіпотези

print("--- 3. Гіпотеза (Закономірність) ---")
print("Факторкільце Zp[x]/(x^2+1) є полем тоді і тільки тоді, коли:")
print(f"Просте число p задовольняє умові: p має остачу 3 (mod 4).")

print("\nДодаткові закономірності з остач (наприклад, для m=8, m=10):")
print(f"Для m=8: p ≡ {results[8]} (mod 8).")
print(f"Для m=10: p ≡ {results[10]} (mod 10).")

# Перевірка умови:
# R = PolynomialRing(GF(last_p), 'x')
# f = R.gen()**2 + 1
# print(f"\nПеревірка: Многочлен x^2 + 1 над GF({last_p}) є незвідним: {f.is_irreducible()}")
```

### 4. Ідеали та факторкільця кільця многочленів

1. Задайте кільце многочленів $R=\mathbb{Z}_{7}[x]$. $\qquad$  %  R.\<x\>=Integers(7)[]
3. Задайте ідеали $I_1=(2x^3+5x+1)$, $I_2=(x^4+3x^3+2)$, $I_3=(x^5+x^3+x+1)$.  $\qquad$    %  I = R.ideal(f(x))
4. Побудуйте факторкільця $R/I$ і знайдіть його порядок. $\qquad$  %  R.quotient(І)
5. Перевірте, для яких ідеалів факторкільце $R/I$ є полем.
6. Перевірте, чи має елемент $(x^5+3x^2+2)+I$ обернений у факторкільці $R/I$ та знайдіть обернений.

```python
# Задання кільця многочленів R = Z7[x]
R.<x> = PolynomialRing(Integers(7), 'x')
print(f"Задане кільце: R = {R}")

# Задання ідеалів
f1 = 2*x^3 + 5*x + 1
f2 = x^4 + 3*x^3 + 2
f3 = x^5 + x^3 + x + 1

I1 = R.ideal(f1)
I2 = R.ideal(f2)
I3 = R.ideal(f3)

print(f"Ідеал I1 = ({f1})")
print(f"Ідеал I2 = ({f2})")
print(f"Ідеал I3 = ({f3})")

print("-" * 60)

## 1. Побудова Факторкілець R/I та Знаходження Їх Порядку

# Факторкільце R/I1
Q1 = R.quotient(I1)
# Порядок факторкільця |R/I| = |Z_p|^(deg(f)), якщо I = (f)
order_Q1 = 7^f1.degree() # 7^3
print(f"Факторкільце R/I1: {Q1} (Порядок: {order_Q1})")

# Факторкільце R/I2
Q2 = R.quotient(I2)
order_Q2 = 7^f2.degree() # 7^4
print(f"Факторкільце R/I2: {Q2} (Порядок: {order_Q2})")

# Факторкільце R/I3
Q3 = R.quotient(I3)
order_Q3 = 7^f3.degree() # 7^5
print(f"Факторкільце R/I3: {Q3} (Порядок: {order_Q3})")

print("-" * 60)

## 2. Перевірка, для яких ідеалів факторкільце R/I є полем

# R/I є полем тоді і тільки тоді, коли ідеал I є максимальним.
# У кільці многочленів над полем (Z_p[x]) ідеал I=(f(x)) є максимальним 
# тоді і тільки тоді, коли многочлен f(x) є незвідним (irreducible) над Z_p.

is_field_Q1 = f1.is_irreducible()
is_field_Q2 = f2.is_irreducible()
is_field_Q3 = f3.is_irreducible()

print(f"R/I1 є полем? (f1={f1} незвідний): {is_field_Q1}")
print(f"R/I2 є полем? (f2={f2} незвідний): {is_field_Q2}")
print(f"R/I3 є полем? (f3={f3} незвідний): {is_field_Q3}")

print("-" * 60)

## 3. Перевірка наявності оберненого елемента та його знаходження

# Елемент a + I має обернений у факторкільці R/I тоді і тільки тоді, коли 
# gcd(a, f) = 1 (де I=(f)).

a = x^5 + 3*x^2 + 2

print(f"Елемент для перевірки: a = {a}")
print("-" * 30)

# Перевірка в Q1 = R/I1
print("Перевірка в R/I1:")
g1 = gcd(a, f1)
has_inverse_Q1 = (g1 == 1)
print(f"НСД(a, f1): {g1}")

if has_inverse_Q1:
    print("-> Обернений ЕЛЕМЕНТ ІСНУЄ.")
    try:
        a_in_Q1 = Q1(a)
        inv_a_Q1 = a_in_Q1.inverse()
        print(f"Обернений елемент: {inv_a_Q1} + I1")
    except Exception as e:
        print("Помилка при обчисленні оберненого.")
else:
    print("-> Обернений ЕЛЕМЕНТ НЕ ІСНУЄ (НСД ≠ 1).")

print("-" * 30)

# Перевірка в Q2 = R/I2
print("Перевірка в R/I2:")
g2 = gcd(a, f2)
has_inverse_Q2 = (g2 == 1)
print(f"НСД(a, f2): {g2}")

if has_inverse_Q2:
    print("-> Обернений ЕЛЕМЕНТ ІСНУЄ.")
    try:
        a_in_Q2 = Q2(a)
        inv_a_Q2 = a_in_Q2.inverse()
        print(f"Обернений елемент: {inv_a_Q2} + I2")
    except Exception as e:
        print("Помилка при обчисленні оберненого.")
else:
    print("-> Обернений ЕЛЕМЕНТ НЕ ІСНУЄ (НСД ≠ 1).")

print("-" * 30)

# Перевірка в Q3 = R/I3
print("Перевірка в R/I3:")
g3 = gcd(a, f3)
has_inverse_Q3 = (g3 == 1)
print(f"НСД(a, f3): {g3}")

if has_inverse_Q3:
    print("-> Обернений ЕЛЕМЕНТ ІСНУЄ.")
    try:
        a_in_Q3 = Q3(a)
        inv_a_Q3 = a_in_Q3.inverse()
        print(f"Обернений елемент: {inv_a_Q3} + I3")
        #xbar — це позначення для класу лишку многочлена x у факторкільці
    except Exception as e:
        print("Помилка при обчисленні оберненого.")
else:
    print("-> Обернений ЕЛЕМЕНТ НЕ ІСНУЄ (НСД ≠ 1).")
```
